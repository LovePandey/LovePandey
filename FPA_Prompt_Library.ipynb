# 🏦 FP&A Prompt Library

This notebook contains reusable AI prompts for Finance workflows.
Each prompt is battle-tested and ready to drop into Claude, ChatGPT, or any LLM.

**Prompts in this library:**
1. Variance Commentary (P&L)
2. Labour Cost Commentary
3. Board Summary Generator
4. Forecast Assumptions Writer
5. KPI Alert Narrative

```python
# ============================================================
# PROMPT 1: Variance Commentary Generator
# Use: Paste your P&L table into the data variable
#      then copy the full prompt into Claude or ChatGPT
# ============================================================

variance_prompt = """
You are a Senior FP&A Analyst preparing a board-level commentary.

Here is the monthly P&L data:

[PASTE YOUR TABLE HERE]

Write a CFO-ready variance commentary. Format:
- 3 bullet points maximum
- Each bullet: what happened, why, what action is recommended
- British English, formal tone
- No filler phrases
- End with one plain-English sentence for non-finance board members
"""

print("✅ Prompt 1 loaded: Variance Commentary")
print("-" * 50)
print(variance_prompt)
```

```python
# ============================================================
# PROMPT 2: Labour Cost Commentary Generator
# Use: Replace table data with your actual headcount numbers
# ============================================================

labour_prompt = """
You are a Senior FP&A Analyst responsible for labour cost reporting.

Here is headcount and cost data:

[PASTE YOUR HEADCOUNT TABLE HERE]

Write a labour cost commentary for the CFO. Include:
- Overall headcount and cost variance summary
- One insight per department flagging a risk or opportunity
- A closing recommendation on hiring vs cost control
- British English, no more than 200 words total
"""

print("✅ Prompt 2 loaded: Labour Cost Commentary")
print("-" * 50)
print(labour_prompt)
```

```python
# ============================================================
# PROMPT 3: Board Summary Generator
# Use: Feed any financial narrative and get exec summary
# ============================================================

board_summary_prompt = """
You are a CFO preparing a one-page executive summary for a board meeting.

Here is the financial narrative for the period:

[PASTE YOUR NARRATIVE HERE]

Rewrite this as a board-ready executive summary. Rules:
- Maximum 150 words
- Start with the single most important number
- Use the structure: Performance → Risk → Outlook
- British English, no jargon
- Last sentence must be a clear recommendation or decision required
"""

print("✅ Prompt 3 loaded: Board Summary Generator")
print("-" * 50)
print(board_summary_prompt)
```

```python
# ============================================================
# MASTER PROMPT SELECTOR
# Run this cell to see all available prompts
# ============================================================

prompts = {
    "1": ("Variance Commentary", variance_prompt),
    "2": ("Labour Cost Commentary", labour_prompt),
    "3": ("Board Summary Generator", board_summary_prompt),
}

print("=" * 50)
print("   📚 FP&A PROMPT LIBRARY — Love Pandey")
print("=" * 50)
for key, (name, _) in prompts.items():
    print(f"  Prompt {key}: {name}")
print("=" * 50)
print("To use: print(prompts['1'][1])  ← change number")
```
