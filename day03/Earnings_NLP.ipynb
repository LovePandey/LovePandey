```python
# ============================================================
# DAY 2: Automated Variance Analysis
# We store P&L data as a Python dictionary (like a table)
# ============================================================

# Each line item has 3 values: budget, actual, variance
pl_data = {
    "Revenue":          {"budget": 5200, "actual": 4850},
    "COGS":             {"budget": 2080, "actual": 2100},
    "Gross Profit":     {"budget": 3120, "actual": 2750},
    "Operating Costs":  {"budget": 1800, "actual": 1650},
    "EBITDA":           {"budget": 1320, "actual": 1100},
}

# Python calculates the variance automatically
for item, values in pl_data.items():
    values["variance"] = values["actual"] - values["budget"]
    values["variance_pct"] = round((values["variance"] / values["budget"]) * 100, 1)

print("✅ P&L data loaded and variances calculated")
print("-" * 45)
print(f"{'Line Item':<20} {'Budget':>8} {'Actual':>8} {'Variance':>10} {'%':>6}")
print("-" * 45)
for item, v in pl_data.items():
    print(f"{item:<20} {v['budget']:>8,} {v['actual']:>8,} {v['variance']:>+10,} {v['variance_pct']:>+5.1f}%")
```

```python
# ============================================================
# CELL 2: Automatic variance flagging
# Python reads the numbers and decides: good, bad, or watch
# ============================================================

def flag_variance(item, variance_pct):
    """
    This is a function — our vending machine.
    Put in: line item name + variance %
    Get out: a flag (🔴 Bad / 🟢 Good / 🟡 Watch)
    """

    # Revenue going down is always bad
    if item == "Revenue" and variance_pct < 0:
        return "🔴 RISK"

    # Costs going UP is bad (we spent more than planned)
    if item in ["COGS", "Operating Costs"] and variance_pct > 0:
        return "🔴 RISK"

    # EBITDA missing is always flagged
    if item == "EBITDA" and variance_pct < 0:
        return "🔴 RISK"

    # Costs going DOWN is good (we saved money)
    if item in ["COGS", "Operating Costs"] and variance_pct < 0:
        return "🟢 POSITIVE"

    # Within 2% either way = fine
    if abs(variance_pct) <= 2:
        return "🟢 ON TRACK"

    # Everything else = watch it
    return "🟡 WATCH"

# Now run every line item through the function
print("✅ Variance Flags Generated")
print("-" * 55)
print(f"{'Line Item':<20} {'Variance %':>10} {'Flag':>20}")
print("-" * 55)
for item, v in pl_data.items():
    flag = flag_variance(item, v["variance_pct"])
    print(f"{item:<20} {v['variance_pct']:>+9.1f}% {flag:>20}")
```

```python
# ============================================================
# CELL 3: Automatic commentary generator
# ============================================================

def generate_commentary(item, values):
    budget = values["budget"]
    actual = values["actual"]
    variance = values["variance"]
    pct = values["variance_pct"]

    if item == "Revenue":
        if variance < 0:
            return f"Revenue underperformed budget by €{abs(variance):,}k ({pct:+.1f}%), likely driven by delayed deal closures. Sales pipeline should be reviewed immediately."
        else:
            return f"Revenue exceeded budget by €{abs(variance):,}k ({pct:+.1f}%), reflecting strong commercial momentum. Ensure sustainability into next period."

    if item == "COGS":
        if variance > 0:
            return f"COGS ran €{abs(variance):,}k over budget ({pct:+.1f}%), compressing gross margin. Supplier pricing and production efficiency should be reviewed."
        else:
            return f"COGS came in €{abs(variance):,}k below budget ({pct:+.1f}%), improving gross margin. Confirm whether savings are structural or a timing benefit."

    if item == "Operating Costs":
        if variance < 0:
            return f"Operating costs were €{abs(variance):,}k below budget ({pct:+.1f}%), likely reflecting headcount underspend. Assess whether underspend creates delivery risk."
        else:
            return f"Operating costs exceeded budget by €{abs(variance):,}k ({pct:+.1f}%). Identify whether overspend is one-off or structural before approving further commitments."

    if item == "EBITDA":
        if variance < 0:
            return f"EBITDA missed budget by €{abs(variance):,}k ({pct:+.1f}%), driven by revenue shortfall and margin pressure. Management action required to protect full-year targets."
        else:
            return f"EBITDA of €{actual:,}k exceeded budget by €{abs(variance):,}k ({pct:+.1f}%), reflecting cost discipline across the business."

    direction = "above" if variance > 0 else "below"
    return f"{item} came in €{abs(variance):,}k {direction} budget ({pct:+.1f}%). Monitor closely in the coming period."


# ---- Generate the full report ----
print("=" * 60)
print("   📋 AUTOMATED VARIANCE COMMENTARY — MARCH 2025")
print("=" * 60)

red_flags = []

for item, values in pl_data.items():
    flag = flag_variance(item, values["variance_pct"])
    commentary = generate_commentary(item, values)
    print(f"\n{flag}  {item}")
    print(f"   {commentary}")
    if "RISK" in flag:
        red_flags.append(item)

print("\n" + "=" * 60)
print("📌 EXECUTIVE SUMMARY")
print("=" * 60)

ebitda = pl_data["EBITDA"]
print(f"\nEBITDA came in at €{ebitda['actual']:,}k against a budget of €{ebitda['budget']:,}k — a shortfall of €{abs(ebitda['variance']):,}k ({ebitda['variance_pct']:+.1f}%).")

if red_flags:
    print(f"Key risk areas requiring CFO attention: {', '.join(red_flags)}.")

print("\nIn plain English: we missed our profit target this month.")
print("The primary driver was a revenue gap that cost savings only partially offset.")
```

```python
# ============================================================
# CELL 4: YOUR REUSABLE VARIANCE TOOL
# Change the numbers below = entire report regenerates
# This is now YOUR tool. Use it every month-end.
# ============================================================

def run_variance_report(data, period="Current Period"):
    """
    THE VENDING MACHINE:
    Put in: any P&L dictionary + period name
    Get out: full automated variance report
    """
    # Calculate variances
    for item, values in data.items():
        values["variance"] = values["actual"] - values["budget"]
        values["variance_pct"] = round((values["variance"] / values["budget"]) * 100, 1)

    # Print report
    print("=" * 60)
    print(f"   📋 VARIANCE REPORT — {period.upper()}")
    print("=" * 60)

    red_flags = []
    for item, values in data.items():
        flag = flag_variance(item, values["variance_pct"])
        commentary = generate_commentary(item, values)
        print(f"\n{flag}  {item}")
        print(f"   {commentary}")
        if "RISK" in flag:
            red_flags.append(item)

    ebitda = data.get("EBITDA", {})
    if ebitda:
        print("\n" + "=" * 60)
        print("📌 EXECUTIVE SUMMARY")
        print("=" * 60)
        print(f"\nEBITDA: €{ebitda['actual']:,}k actual vs €{ebitda['budget']:,}k budget ({ebitda['variance_pct']:+.1f}%).")
        if red_flags:
            print(f"CFO attention required: {', '.join(red_flags)}.")

    print("\n" + "=" * 60)
    print("✅ Report complete. Copy above into your board pack.")
    print("=" * 60)


# ============================================================
# 👇 THIS IS WHERE YOU CHANGE YOUR DATA EACH MONTH
# Just update the budget and actual numbers below
# ============================================================

april_data = {
    "Revenue":         {"budget": 5500, "actual": 4900 },
    "COGS":            {"budget": 2200, "actual": 2180},
    "Gross Profit":    {"budget": 3300, "actual": 3570},
    "Operating Costs": {"budget": 1850, "actual": 1900},
    "EBITDA":          {"budget": 1450, "actual": 1670},
}

# Run it — one line does everything
run_variance_report(april_data, period="April 2025")
```

```python
# ============================================================
# CELL 2: Load earnings call transcripts
# These are realistic CFO/CEO quotes — typical of real calls
# ============================================================

# We have TWO companies to compare — one strong, one struggling
company_a = """
We delivered exceptional revenue growth of 18 percent this quarter,
exceeding our guidance and demonstrating strong momentum across all segments.
Customer retention remains outstanding and we are confident in our ability
to achieve full year targets. Our pipeline is robust and expanding rapidly.
The leadership team is energised and we see significant opportunities ahead.
We are pleased to raise our full year guidance and remain optimistic about
our market position. Demand is accelerating and margins are improving.
Cash generation was strong and we continue to invest in future growth.
"""

company_b = """
Revenue came in below expectations this quarter due to challenging market
conditions and prolonged sales cycles. We are cautious about the near term
outlook and have revised our guidance downward. Headwinds from inflation and
rising costs continue to pressure margins. Customer churn has increased and
we are working to address retention issues. The macro environment remains
uncertain and we expect continued pressure in the coming quarters.
We are implementing cost reduction measures and restructuring certain business
units. Management acknowledges the difficult operating environment ahead.
"""

print("✅ Two earnings call transcripts loaded")
print(f"📄 Company A: {len(company_a.split())} words")
print(f"📄 Company B: {len(company_b.split())} words")
print("\nReady for sentiment analysis...")
```

```python
# ============================================================
# CELL 3: Sentiment Analyser
# TextBlob scores every sentence from -1.0 to +1.0
# -1.0 = very negative | 0.0 = neutral | +1.0 = very positive
# ============================================================

import nltk
nltk.download('punkt')

def analyse_earnings(company_name, transcript):
    """
    Put in: company name + transcript text
    Get out: full sentiment analysis report
    """
    blob = TextBlob(transcript)

    # Overall sentiment scores
    polarity = round(blob.sentiment.polarity, 3)        # positive/negative
    subjectivity = round(blob.sentiment.subjectivity, 3) # factual vs opinion

    # Score every sentence individually
    sentence_scores = []
    for sentence in blob.sentences:
        score = round(sentence.sentiment.polarity, 3)
        sentence_scores.append((str(sentence).strip(), score))

    # Find most positive and most negative sentence
    most_positive = max(sentence_scores, key=lambda x: x[1])
    most_negative = min(sentence_scores, key=lambda x: x[1])

    # Overall verdict
    if polarity > 0.2:
        verdict = "🟢 BULLISH"
    elif polarity > 0.0:
        verdict = "🟡 CAUTIOUSLY POSITIVE"
    elif polarity > -0.2:
        verdict = "🟡 CAUTIOUSLY NEGATIVE"
    else:
        verdict = "🔴 BEARISH"

    # Print report
    print("=" * 60)
    print(f"  📊 SENTIMENT REPORT — {company_name.upper()}")
    print("=" * 60)
    print(f"\n  Overall Verdict:    {verdict}")
    print(f"  Polarity Score:     {polarity:+.3f}  (range: -1.0 to +1.0)")
    print(f"  Subjectivity Score: {subjectivity:.3f}   (0=factual, 1=opinion)")

    print(f"\n  📈 Most Positive Sentence (score: {most_positive[1]:+.3f}):")
    print(f"  \"{most_positive[0][:100]}...\"")

    print(f"\n  📉 Most Negative Sentence (score: {most_negative[1]:+.3f}):")
    print(f"  \"{most_negative[0][:100]}...\"")

    print("\n  📋 All Sentence Scores:")
    for i, (sent, score) in enumerate(sentence_scores, 1):
        bar = "🟢" if score > 0.1 else "🔴" if score < -0.1 else "⚪"
        print(f"  {bar} S{i}: {score:+.3f} | {sent[:70]}...")

    print()
    return polarity

# Run both companies
score_a = analyse_earnings("Company A", company_a)
score_b = analyse_earnings("Company B", company_b)
```

```python
# ============================================================
# CELL 4: Finance Risk Word Scanner
# TextBlob misses domain-specific language
# This cell adds a Finance lens on top
# ============================================================

# Words that signal RISK in earnings calls
# (TextBlob doesn't know these are negative in finance context)
risk_words = [
    "headwinds", "cautious", "challenging", "uncertain", "revised downward",
    "below expectations", "pressure", "churn", "restructuring", "declining",
    "difficult", "concerns", "slowdown", "weakness", "volatile", "risk",
    "reduce", "cut", "loss", "warning", "miss", "disappointing"
]

# Words that signal CONFIDENCE
positive_words = [
    "exceptional", "strong", "confident", "robust", "momentum", "growth",
    "exceeded", "outperformed", "optimistic", "expanding", "accelerating",
    "raising guidance", "record", "outstanding", "energised", "improving",
    "opportunities", "pleased", "delivering", "ahead of"
]

def finance_sentiment_scan(company_name, transcript):
    """
    Finance-aware sentiment scanner
    Counts risk vs confidence signals specifically
    """
    text_lower = transcript.lower()

    # Count hits for each category
    risk_hits = []
    positive_hits = []

    for word in risk_words:
        count = text_lower.count(word)
        if count > 0:
            risk_hits.append((word, count))

    for word in positive_words:
        count = text_lower.count(word)
        if count > 0:
            positive_hits.append((word, count))

    # Sort by frequency
    risk_hits.sort(key=lambda x: x[1], reverse=True)
    positive_hits.sort(key=lambda x: x[1], reverse=True)

    total_risk = sum(c for _, c in risk_hits)
    total_positive = sum(c for _, c in positive_hits)

    # Finance verdict
    ratio = total_positive / (total_risk + 1)  # +1 avoids division by zero
    if ratio > 2:
        finance_verdict = "🟢 INVEST SIGNAL — Management is confident"
    elif ratio > 1:
        finance_verdict = "🟡 HOLD SIGNAL — Mixed signals, monitor closely"
    else:
        finance_verdict = "🔴 CAUTION SIGNAL — Risk language dominates"

    print("=" * 60)
```

```python
# ============================================================
# Run the finance scanner on both companies
# ============================================================

pos_a, risk_a = finance_sentiment_scan("Company A", company_a)
pos_b, risk_b = finance_sentiment_scan("Company B", company_b)

# Head to head
print("=" * 60)
print("  ⚔️  HEAD TO HEAD COMPARISON")
print("=" * 60)
print(f"\n  {'Metric':<25} {'Company A':>12} {'Company B':>12}")
print(f"  {'-'*49}")
print(f"  {'Confidence signals':<25} {pos_a:>12} {pos_b:>12}")
print(f"  {'Risk signals':<25} {risk_a:>12} {risk_b:>12}")
print(f"  {'Signal ratio':<25} {pos_a/(risk_a+1):>11.2f}x {pos_b/(risk_b+1):>11.2f}x")
print(f"\n  📌 CONCLUSION:")
if pos_a > pos_b:
    print(f"  Company A = 🟢 INVEST SIGNAL")
    print(f"  Company B = 🔴 CAUTION SIGNAL")
    print(f"  A quant analyst would flag Company B for review.")
```
